```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
import seaborn as sns

df = pd.read_csv("Mall_Customers.csv")

plt.figure(figsize=(10, 7))

# Plot all clusters except noise
clusters = df[df["cluster"] != -1]
noise = df[df["cluster"] == -1]

plt.scatter(
    clusters["Annual Income (k$)"],
    clusters["Spending Score (1-100)"],
    c=clusters["cluster"],
    cmap="viridis",
    s=120,
    edgecolor="black",
    alpha=0.8,
    label="Clusters"
)

# Plot noise points separately
if len(noise) > 0:
    plt.scatter(
        noise["Annual Income (k$)"],
        noise["Spending Score (1-100)"],
        color="red",
        s=140,
        marker="X",
        edgecolor="black",
        label="Noise (-1)"
    )

plt.title("DBSCAN Clustering of Mall Customers", fontsize=18, fontweight="bold")
plt.xlabel("Annual Income (k$)", fontsize=14)
plt.ylabel("Spending Score (1-100)", fontsize=14)

plt.grid(True, linestyle="--", alpha=0.4)

cbar = plt.colorbar()
cbar.set_label("Cluster ID")

plt.legend()
plt.tight_layout()
plt.show()

# Cluster Summary
print("\nCluster Counts")
print("-" * 30)
print(df["cluster"].value_counts().sort_index())

print("\nTotal Clusters (excluding noise):", len(set(df["cluster"])) - (1 if -1 in df["cluster"].values else 0))
print("Noise Points:", (df["cluster"] == -1).sum())
```
