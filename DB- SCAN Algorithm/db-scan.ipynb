```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
import seaborn as sns

sns.set()

url = "https://raw.githubusercontent.com/Pankaj-Str/Complete-Python-Mastery/refs/heads/main/53%20DataSet/Mall_Customers.csv"
df = pd.read_csv(url)

print("Dataset Loaded Successfully!")
print(df.head())

X = df[["Annual Income (k$)", "Spending Score (1-100)"]].values

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Apply DBscan

dbscan = DBSCAN(
    eps = 0.5,
    min_samples = 5,
)

lables = dbscan.fit_predict(X_scaled)

df["cluster"] = lables
```

```python
plt.figure(figsize=(8,6))
plt.scatter(df["Annual Income (k$)"], df["Spending Score (1-100)"],
            c=df["cluster"], cmap="viridis", s=60)

plt.title("DBSCAN Clustering (Mall Customers)")
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score")
plt.show()

print(df["cluster"].value_counts())
```
